```sql
--  Required IDs
SELECT *
FROM oulad.oulad_silver.student_assessment_silver
WHERE id_assessment IS NULL
   OR id_student IS NULL;
```

```sql
-- Duplicate Grain

SELECT
    id_student,
    id_assessment,
    COUNT(*) AS row_count
FROM oulad.oulad_silver.student_assessment_silver
GROUP BY
    id_student,
    id_assessment
HAVING COUNT(*) > 1;
```

```sql
-- is_banked Validity
SELECT *
FROM oulad.oulad_silver.student_assessment_silver
WHERE is_banked NOT IN (0, 1)
   OR is_banked IS NULL;
```

```sql
-- Score Range
SELECT *
FROM oulad.oulad_silver.student_assessment_silver
WHERE score < 0
   OR score > 100;
```

```sql
-- REVIEW: Negative date_submitted MEANING day submitted early
SELECT *
FROM oulad.oulad_silver.student_assessment_silver
WHERE date_submitted < 0;
```

```sql
-- NULL date_submitted
SELECT *
FROM oulad.oulad_silver.student_assessment_silver
WHERE date_submitted IS NULL;
```

```sql
SELECT
    COUNT(*) AS total_rows,
    COUNT_IF(id_assessment IS NULL) AS null_id_assessment,
    COUNT_IF(id_student IS NULL) AS null_id_student,
    COUNT_IF(TRY_CAST(id_assessment AS INT) IS NULL
             AND id_assessment IS NOT NULL) AS invalid_id_assessment,
    COUNT_IF(TRY_CAST(id_student AS INT) IS NULL
             AND id_student IS NOT NULL) AS invalid_id_student,
    COUNT_IF(TRY_CAST(date_submitted AS INT) IS NULL
             AND date_submitted IS NOT NULL) AS invalid_date_submitted,
    COUNT_IF(TRY_CAST(is_banked AS INT) IS NULL
             AND is_banked IS NOT NULL) AS invalid_is_banked
FROM oulad.oulad_bronze.student_assessment_bronze;
```

```sql
-- ? AND NULL Score Profiling
SELECT
    COUNT(*) AS total_rows,
    COUNT_IF(score IS NULL) AS null_scores,
    COUNT_IF(TRIM(score) = '?') AS question_mark_scores,
    COUNT_IF(
        score IS NOT NULL
        AND TRIM(score) <> '?'
        AND TRY_CAST(TRIM(score) AS DOUBLE) IS NULL
    ) AS invalid_scores
FROM oulad.oulad_bronze.student_assessment_bronze;
```

```sql
SELECT
    COUNT(*) AS null_score_count
FROM oulad.oulad_silver.student_assessment_silver
WHERE score IS NULL;
```

```sql
--  Ingestion Metadata
SELECT
    COUNT(*) AS total_rows,
    COUNT_IF(ingestion_timestamp IS NULL) AS null_ingestion_timestamp,
    COUNT_IF(ingestion_date IS NULL) AS null_ingestion_date
FROM oulad.oulad_silver.student_assessment_silver;
```

```sql
-- COMPARE BRONZE AND SILVER ROWS
SELECT
    (SELECT COUNT(*)
     FROM oulad.oulad_bronze.student_assessment_bronze) AS bronze_rows,

    (SELECT COUNT(*)
     FROM oulad.oulad_silver.student_assessment_silver) AS silver_rows;
```

```sql
-- Count negative date_submitted values
-- REVIEW: 2,057 records have negative date_submitted values, with a minimum of -11. These values were retained because date_submitted is a relative day offset. Further investigation is required to determine whether these records are expected based on the source data.

SELECT
    COUNT(*) AS negative_date_count,
    MIN(date_submitted) AS minimum_date_submitted
FROM oulad.oulad_silver.student_assessment_silver
WHERE date_submitted < 0;
```
